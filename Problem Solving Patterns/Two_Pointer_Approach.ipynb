# Two Pointer Approach

---

## 🔍 **Definition**

The **Two Pointer Approach** is a problem-solving technique where **two indices (pointers)** are used to traverse a data structure (usually an array or string), either:

✔ from both ends
✔ or moving independently in the same direction

to solve the problem **more efficiently than brute force**.

---

### 🧠 **Intuition**

Instead of using nested loops (O(n²)),
we **move two pointers intelligently** to reduce work to **O(n)**.

Think of it like:

👉 One pointer at the **start**
👉 One pointer at the **end**
👉 Move them based on conditions

---

## 🖼 **Simple Visual Diagram**

### Example: Two pointers from both ends

```
Array:  [ 2   5   7   9   11 ]
Index:   0   1   2   3    4
         ^               ^
         L               R
```

* `L` → Left Pointer (starts at beginning)
* `R` → Right Pointer (starts at end)

As the algorithm runs:

```
Step 1:
L -> 2
R -> 11

Step 2:
L moves →
R moves ←
```

Pointers move **towards each other** until they meet.

# 🧮 Example 1 — Check if an Array is a Palindrome

### 📝 Problem

Check if array reads the same forward & backward.

### 💡 Idea

Compare first & last element, then move inward.

---



````python
def is_palindrome(arr):
    left = 0
    right = len(arr) - 1

    while left < right:
        if arr[left] != arr[right]:
            return False
        
        left += 1
        right -= 1
    
    return True

# Example
arr = [1, 2, 3, 2, 1]
print(is_palindrome(arr)) 
````

# 🎯 Example 2 — Two Sum in Sorted Array (Two Pointer Version)

### 📝 Problem

Find two numbers whose sum equals target
(array is sorted)

---

### 💡 Logic

```
If sum < target → move left pointer →
If sum > target → move right pointer ←
If sum == target → found
```

---

### ⏱ Time Complexity

**O(n)**
instead of
❌ brute force O(n²)

````python
def two_sum_sorted(arr, target):
    left = 0
    right = len(arr) - 1

    while left < right:
        curr = arr[left] + arr[right]

        if curr == target:
            return (left, right)
        elif curr < target:
            left += 1
        else:
            right -= 1

    return None

# Example
arr = [1, 2, 4, 7, 11, 15]
target = 15
print(two_sum_sorted(arr, target))
````

# 🧩 Example 3 — Remove Duplicates From Sorted Array

### 📝 Problem

Modify array to keep unique elements only.

---

### 💡 Idea

Use:

✔ slow pointer — keeps position of last unique
✔ fast pointer — scans ahead

---

### 🖼 Diagram

```
Index:   0 1 2 3 4 5
Array:  [1 1 2 2 3 3]

slow -> 1
fast -> 1
```

Move fast ahead
Update slow when new value found

---

````python
def remove_duplicates(arr):
    if not arr:
        return 0
    
    slow = 0

    for fast in range(1, len(arr)):
        if arr[fast] != arr[slow]:
            slow += 1
            arr[slow] = arr[fast]

    return slow + 1

# Example
arr = [1,1,2,2,3,3]
length = remove_duplicates(arr)
print(length)            # Output: 3
print(arr[:length])      # Output: [1,2,3]
````
